```python
# 异常检测结果分析工具，这个工具可以帮助你分析异常检测模型的输出结果，特别是test_output.pkl文件中的信息。
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import glob
import json
from datetime import datetime

# %matplotlib inline
plt.rcParams['font.sans-serif'] = ['SimHei']  # 用来正常显示中文标签
plt.rcParams['axes.unicode_minus'] = False  # 用来正常显示负号


def find_model_path(result_path, model_id):
    """
    查找模型路径
    """
    # 检查是否是通用模型路径
    universal_model_path = os.path.join(result_path, 'universal_model')
    if os.path.exists(universal_model_path):
        result_path = universal_model_path
        print(f"检测到通用模型路径，更新为: {result_path}")

    if model_id is None:
        dir_content = os.listdir(result_path)
        subfolders = [subf for subf in dir_content if os.path.isdir(os.path.join(result_path, subf)) and subf != "logs"]
        # 过滤出时间戳格式的目录
        date_times = []
        for subf in subfolders:
            try:
                dt = datetime.strptime(subf, '%d%m%Y_%H%M%S')
                date_times.append(dt)
            except ValueError:
                continue
        if not date_times:
            raise Exception(f"在 {result_path} 中未找到有效的时间戳目录")
        date_times.sort()
        model_datetime = date_times[-1]
        model_id = model_datetime.strftime('%d%m%Y_%H%M%S')

    if str(model_id).startswith('-'):
        # 处理相对ID（如-1, -2等）
        dir_content = os.listdir(result_path)
        subfolders = [subf for subf in dir_content if os.path.isdir(os.path.join(result_path, subf)) and subf != "logs"]
        date_times = []
        datetime_folders = {}
        for subf in subfolders:
            try:
                dt = datetime.strptime(subf, '%d%m%Y_%H%M%S')
                date_times.append(dt)
                datetime_folders[dt] = subf
            except ValueError:
                continue
        if not date_times:
            raise Exception(f"在 {result_path} 中未找到有效的时间戳目录")
        date_times.sort()
        index = int(model_id)
        model_datetime = date_times[index]
        model_id = datetime_folders[model_datetime]

    model_path = os.path.join(result_path, model_id)
    print(f"使用模型路径: {model_path}")
    return model_path


def analyze_entity_results(entity_path, confidence_level, entity_name):
    """
    分析单个实体的结果
    """
    # 加载测试输出
    try:
        test_output = pd.read_pickle(os.path.join(entity_path, "test_output.pkl"))
    except FileNotFoundError:
        print(f"在 {entity_path} 中未找到 test_output.pkl 文件")
        return

    # 获取异常分数
    if 'A_Score_Global' in test_output.columns:
        anomaly_scores = test_output['A_Score_Global'].values
    else:
        print("未找到全局异常分数 'A_Score_Global'")
        return

    print(f"异常分数统计:")
    print(f"  最小值: {np.min(anomaly_scores):.6f}")
    print(f"  最大值: {np.max(anomaly_scores):.6f}")
    print(f"  平均值: {np.mean(anomaly_scores):.6f}")
    print(f"  中位数: {np.median(anomaly_scores):.6f}")
    print(f"  标准差: {np.std(anomaly_scores):.6f}")

    # 计算指定置信区间的阈值
    threshold = np.percentile(anomaly_scores, confidence_level * 100)
    print(f"\n{confidence_level * 100:.1f}% 置信区间阈值: {threshold:.6f}")

    # 检测异常点
    anomalies = np.where(anomaly_scores >= threshold)[0]
    print(f"检测到的异常点数量: {len(anomalies)}")
    if len(anomalies) > 0:
        print(f"异常点位置: {anomalies.tolist()}")

    # 尝试加载Epsilon方法的阈值
    try:
        with open(os.path.join(entity_path, "summary.txt"), "r") as f:
            summary = json.load(f)
            epsilon_threshold = summary.get("epsilon_result", {}).get("threshold", None)
            if epsilon_threshold:
                print(f"Epsilon方法阈值: {epsilon_threshold:.6f}")
                epsilon_anomalies = np.where(anomaly_scores >= epsilon_threshold)[0]
                print(f"Epsilon方法检测到的异常点数量: {len(epsilon_anomalies)}")
    except FileNotFoundError:
        print("未找到 summary.txt 文件")

    # 可视化
    visualize_anomaly_scores(test_output, threshold, confidence_level, entity_name, anomaly_scores, anomalies)

    # 打印详细信息
    print(f"\n实体 {entity_name} 的详细分析:")
    print("-" * 40)

    # 显示前10个最高异常分数的时间点
    top_anomalies = np.argsort(anomaly_scores)[-10:][::-1]
    print("前10个最高异常分数的时间点:")
    for i, idx in enumerate(top_anomalies):
        if 'timestamp' in test_output.columns:
            ts = test_output.iloc[idx]['timestamp']
            print(f"  {i + 1:2d}. 时间戳: {ts}, 位置: {idx:4d}, 分数: {anomaly_scores[idx]:.6f}")
        else:
            print(f"  {i + 1:2d}. 位置: {idx:4d}, 分数: {anomaly_scores[idx]:.6f}")

    # 分析预测误差
    if 'True_0' in test_output.columns and 'Forecast_0' in test_output.columns:
        true_values = test_output['True_0'].values
        forecast_values = test_output['Forecast_0'].values
        forecast_errors = np.abs(true_values - forecast_values)
        
        print(f"\n预测误差统计:")
        print(f"  最小误差: {np.min(forecast_errors):.6f}")
        print(f"  最大误差: {np.max(forecast_errors):.6f}")
        print(f"  平均误差: {np.mean(forecast_errors):.6f}")
        print(f"  中位数误差: {np.median(forecast_errors):.6f}")
        
        # 分析异常点处的预测误差
        if len(anomalies) > 0:
            anomaly_errors = forecast_errors[anomalies]
            print(f"\n异常点处的预测误差:")
            print(f"  最小误差: {np.min(anomaly_errors):.6f}")
            print(f"  最大误差: {np.max(anomaly_errors):.6f}")
            print(f"  平均误差: {np.mean(anomaly_errors):.6f}")


def visualize_anomaly_scores(test_output, threshold, confidence_level, entity_name, anomaly_scores, anomalies):
    """
    可视化异常分数和检测到的异常点
    """
    # 创建图形
    fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(15, 12))

    # 时间轴
    if 'timestamp' in test_output.columns:
        x_axis = test_output['timestamp']
        x_label = '时间戳'
    else:
        x_axis = range(len(anomaly_scores))
        x_label = '时间点'

    # 绘制异常分数
    ax1.plot(x_axis, anomaly_scores, 'b-', linewidth=1, label='异常分数')
    ax1.axhline(y=threshold, color='r', linestyle='--', linewidth=1,
                label=f'{confidence_level * 100:.1f}% 置信区间阈值 ({threshold:.4f})')

    # 标记异常点
    if len(anomalies) > 0:
        ax1.scatter([x_axis[i] for i in anomalies], [anomaly_scores[i] for i in anomalies],
                    color='red', s=20, label=f'检测到的异常点 ({len(anomalies)})', zorder=5)

    ax1.set_xlabel(x_label)
    ax1.set_ylabel('异常分数')
    ax1.set_title(f'实体 {entity_name} 的异常分数和检测结果')
    ax1.legend()
    ax1.grid(True, alpha=0.3)

    # 绘制异常点分布直方图
    ax2.hist(anomaly_scores, bins=50, alpha=0.7, color='blue', edgecolor='black')
    ax2.axvline(x=threshold, color='r', linestyle='--', linewidth=2,
                label=f'{confidence_level * 100:.1f}% 置信区间阈值 ({threshold:.4f})')

    if len(anomalies) > 0:
        ax2.hist([anomaly_scores[i] for i in anomalies], bins=30, alpha=0.7, color='red',
                 label=f'检测到的异常点 ({len(anomalies)})')

    ax2.set_xlabel('异常分数')
    ax2.set_ylabel('频率')
    ax2.set_title(f'实体 {entity_name} 的异常分数分布')
    ax2.legend()
    ax2.grid(True, alpha=0.3)

    # 新增：绘制预测值、真实值和重构值（如果存在）
    if 'True_0' in test_output.columns and 'Forecast_0' in test_output.columns:
        true_values = test_output['True_0'].values
        forecast_values = test_output['Forecast_0'].values
        
        ax3.plot(x_axis, true_values, 'g-', linewidth=1, label='真实值')
        ax3.plot(x_axis, forecast_values, 'b-', linewidth=1, label='预测值')
        
        # 如果存在重构值也一并绘制
        if 'Recon_0' in test_output.columns:
            recon_values = test_output['Recon_0'].values
            ax3.plot(x_axis, recon_values, 'm-', linewidth=1, label='重构值')
        
        # 标记异常点
        if len(anomalies) > 0:
            ax3.scatter([x_axis[i] for i in anomalies], [true_values[i] for i in anomalies],
                        color='red', s=20, label=f'异常点 ({len(anomalies)})', zorder=5)
        
        ax3.set_xlabel(x_label)
        ax3.set_ylabel('值')
        ax3.set_title(f'实体 {entity_name} 的预测值、真实值和重构值对比')
        ax3.legend()
        ax3.grid(True, alpha=0.3)
    else:
        ax3.text(0.5, 0.5, '无预测值数据可显示', ha='center', va='center', transform=ax3.transAxes)
        ax3.set_title(f'实体 {entity_name} 的时间序列数据')

    plt.tight_layout()
    plt.show()


def analyze_anomaly_detection_results(result_path, confidence_level=0.95, model_id='-1'):
    """
    分析异常检测结果，可视化异常分数和检测到的异常点

    Parameters:
    result_path: 结果路径
    confidence_level: 置信区间 (例如 0.95)
    model_id: 模型ID
    """

    # 查找模型路径
    save_path = find_model_path(result_path, model_id)

    # 对于通用模型，需要处理所有实体
    if 'universal_model' in save_path:
        # 获取所有实体目录（数字命名和Cell命名）
        entity_dirs = [d for d in os.listdir(save_path)
                       if os.path.isdir(os.path.join(save_path, d)) and (d.isdigit() or d.startswith('Cell'))]
        
        if entity_dirs:
            # 分别排序数字目录和Cell目录
            numeric_dirs = sorted([d for d in entity_dirs if d.isdigit()], key=int)
            cell_dirs = sorted([d for d in entity_dirs if d.startswith('Cell')])
            
            # 合并目录列表
            all_dirs = numeric_dirs + cell_dirs
            
            # 处理所有实体目录
            for entity in all_dirs:
                entity_path = os.path.join(save_path, entity)
                print(f"\n{'=' * 50}")
                print(f"分析实体 {entity}")
                print(f"{'=' * 50}")
                analyze_entity_results(entity_path, confidence_level, entity)
        else:
            # 如果没有实体子目录，直接分析主路径
            analyze_entity_results(save_path, confidence_level, "main")
    else:
        # 非通用模型直接分析
        analyze_entity_results(save_path, confidence_level, "main")


def compare_entities(result_path, model_id='-1'):
    """
    比较所有实体的异常检测结果
    """
    save_path = find_model_path(result_path, model_id)

    if 'universal_model' not in save_path:
        print("该模型不是通用模型，无法比较多个实体")
        return

    # 获取所有实体目录（数字命名和Cell命名）
    entity_dirs = [d for d in os.listdir(save_path)
                   if os.path.isdir(os.path.join(save_path, d)) and (d.isdigit() or d.startswith('Cell'))]

    if not entity_dirs:
        print("未找到实体目录")
        return

    entity_stats = []
    
    # 分别排序数字目录和Cell目录
    numeric_dirs = sorted([d for d in entity_dirs if d.isdigit()], key=int)
    cell_dirs = sorted([d for d in entity_dirs if d.startswith('Cell')])
    
    # 合并目录列表
    all_dirs = numeric_dirs + cell_dirs

    for entity in all_dirs:
        entity_path = os.path.join(save_path, entity)
        try:
            test_output = pd.read_pickle(os.path.join(entity_path, "test_output.pkl"))
            if 'A_Score_Global' in test_output.columns:
                anomaly_scores = test_output['A_Score_Global'].values
                stats = {
                    'entity': entity,
                    'count': len(anomaly_scores),
                    'min': np.min(anomaly_scores),
                    'max': np.max(anomaly_scores),
                    'mean': np.mean(anomaly_scores),
                    'median': np.median(anomaly_scores),
                    'std': np.std(anomaly_scores)
                }
                entity_stats.append(stats)
        except FileNotFoundError:
            print(f"实体 {entity} 缺少 test_output.pkl 文件")

    if not entity_stats:
        print("没有可比较的实体")
        return

    # 创建比较表
    print("\n各实体异常分数统计比较:")
    print("-" * 80)
    print(f"{'实体':<8} {'数据点数':<10} {'最小值':<12} {'最大值':<12} {'平均值':<12} {'中位数':<12} {'标准差':<12}")
    print("-" * 80)
    for stats in entity_stats:
        print(f"{stats['entity']:<8} {stats['count']:<10} {stats['min']:<12.6f} {stats['max']:<12.6f} "
              f"{stats['mean']:<12.6f} {stats['median']:<12.6f} {stats['std']:<12.6f}")

    # 可视化比较
    entities = [s['entity'] for s in entity_stats]
    means = [s['mean'] for s in entity_stats]
    stds = [s['std'] for s in entity_stats]
    maxs = [s['max'] for s in entity_stats]

    fig, ax = plt.subplots(1, 1, figsize=(12, 6))
    x = np.arange(len(entities))
    width = 0.25

    ax.bar(x - width, means, width, label='平均值', alpha=0.8)
    ax.bar(x, stds, width, label='标准差', alpha=0.8)
    ax.bar(x + width, maxs, width, label='最大值', alpha=0.8)

    ax.set_xlabel('实体')
    ax.set_ylabel('异常分数')
    ax.set_title('各实体异常分数统计比较')
    ax.set_xticks(x)
    ax.set_xticklabels(entities)
    ax.legend()
    ax.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()


def visualize_entity_trends(dataset_path, model_path, confidence_level=0.95):
    """
    可视化训练集和测试集中各实体的趋势，并标出异常点
    """
    # 获取训练实体和测试实体
    if 'CALCE2' in dataset_path:
        from utils import get_calce2_train_test_splits, load_calce2_entity_data
        train_entities, test_entities = get_calce2_train_test_splits()
        load_func = load_calce2_entity_data
    else:  # CALCE
        from utils import get_calce_train_test_splits, load_calce_entity_data
        train_entities, test_entities = get_calce_train_test_splits()
        load_func = load_calce_entity_data

    print(f"训练实体: {train_entities}")
    print(f"测试实体: {test_entities}")

    # 创建图表
    fig, axes = plt.subplots(2, 1, figsize=(15, 12))

    # 绘制训练实体趋势
    print("\n加载训练实体数据...")
    for entity in train_entities:
        try:
            (train_data, _), (_, _) = load_func(entity)
            if train_data is not None:
                # 只绘制第一个特征
                if len(train_data.shape) > 1:
                    data = train_data[:, 0]
                else:
                    data = train_data
                axes[0].plot(data, label=f'实体 {entity}', alpha=0.7)
        except Exception as e:
            print(f"加载实体 {entity} 数据时出错: {e}")

    axes[0].set_title('训练实体趋势')
    axes[0].set_xlabel('时间点')
    axes[0].set_ylabel('值')
    axes[0].legend()
    axes[0].grid(True, alpha=0.3)

    # 绘制测试实体趋势和异常点
    print("\n加载测试实体数据和异常检测结果...")
    model_datetime = os.path.basename(model_path)

    for entity in test_entities:
        try:
            # 加载测试数据
            (_, _), (test_data, _) = load_func(entity)
            if test_data is not None:
                # 只绘制第一个特征
                if len(test_data.shape) > 1:
                    data = test_data[:, 0]
                else:
                    data = test_data
                axes[1].plot(data, label=f'实体 {entity} (测试)', alpha=0.7)

                # 加载异常检测结果
                # 根据实体名称类型构建路径
                if isinstance(entity, int) or (isinstance(entity, str) and entity.isdigit()):
                    entity_dir = str(entity)
                else:
                    entity_dir = entity
                    
                entity_result_path = os.path.join(model_path, entity_dir)
                if os.path.exists(entity_result_path):
                    test_output_path = os.path.join(entity_result_path, "test_output.pkl")
                    if os.path.exists(test_output_path):
                        test_output = pd.read_pickle(test_output_path)
                        if 'A_Score_Global' in test_output.columns:
                            anomaly_scores = test_output['A_Score_Global'].values
                            threshold = np.percentile(anomaly_scores, confidence_level * 100)
                            anomalies = np.where(anomaly_scores >= threshold)[0]

                            # 标出异常点
                            if len(anomalies) > 0:
                                axes[1].scatter(anomalies, [data[i] for i in anomalies],
                                                marker='o', s=50,
                                                label=f'实体 {entity} 异常点 ({len(anomalies)})',
                                                zorder=5)
        except Exception as e:
            print(f"处理实体 {entity} 时出错: {e}")

    axes[1].set_title(f'测试实体趋势和异常点 (置信度: {confidence_level * 100:.1f}%)')
    axes[1].set_xlabel('时间点')
    axes[1].set_ylabel('值')
    axes[1].legend()
    axes[1].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()


## 使用示例，以下是如何使用这个分析工具的示例：
# 分析CALCE数据集结果
result_path = './output/CALCE2'
print(f"分析路径: {result_path}")

# 分析异常检测结果
analyze_anomaly_detection_results(result_path, confidence_level=0.9)
# 比较各实体结果
compare_entities(result_path)
# 可视化训练集和测试集中各实体的趋势及异常点
# 注意：请根据实际情况调整以下路径
dataset_path = './datasets/CALCE2'  # 数据集路径
model_path = './output/CALCE2/universal_model/25112025_142327'  # 模型结果路径
visualize_entity_trends(dataset_path, model_path, confidence_level=0.9)
```
