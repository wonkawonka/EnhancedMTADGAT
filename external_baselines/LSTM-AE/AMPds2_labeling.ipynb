```python
import pandas as pd
import torch
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime
import seaborn as sns
from matplotlib import rc
from pylab import rcParams
%matplotlib inline
```

```python
elec_df = pd.read_csv('./Electricity_WHE.csv')
water_df = pd.read_csv('./Water_WHW.csv')
gas_df = pd.read_csv('./NaturalGas_WHG.csv')
```

```python
data = pd.merge(elec_df, water_df, on="unix_ts")
data = data.drop(['APF','V', 'I', 'f', 'DPF', 'Pt', 'Q', 'Qt', 'S', 'St', 'counter', 'inst_rate'], axis=1)
data = pd.merge(data, gas_df, on="unix_ts")
data = data.drop(['counter', 'inst_rate'], axis=1)
data.columns=['date', 'sensor_P', 'sensor_W', 'sensor_G']

data['date'] = data['date'].astype(int)
data['date'] = data['date'].apply(lambda x: datetime.utcfromtimestamp(x).strftime('%Y-%m-%d %H:%M:%S'))
df = data.copy()
df['date'] = pd.to_datetime(df['date'])
normal_df = df.set_index('date')
normal_df
```

```python
sns.set(rc={'figure.figsize':(11, 4)})    
```

```python
an_idx_list = []
for i in range(105):
    anomaly_idx = np.where(normal_df.iloc[i*10080:(i+1)*10080]['sensor_P']>3950)
    anomaly_idx = anomaly_idx[0]+ (i*10080)
    an_idx_list.extend(anomaly_idx)
    idx = normal_df.iloc[anomaly_idx].index
    #ax_p=normal_df.loc['2012-04-01':'2012-04-08-07','sensor_P'].plot(linewidth=0.5);
    ax_p=normal_df.iloc[i*10080:(i+1)*10080]['sensor_P'].plot(linewidth=0.5);
    ax_p.set_title('Electricity');
    ax_p.set_ylabel('Meters (VAR)');
    ax_p.set_xlabel('Time');
    ax_p = plt.scatter(idx,normal_df.loc[idx,'sensor_P'],c='r')
    print(f"Number of anomaly : {len(idx)} Out of 10080")
    plt.show()
    plt.close()
```

```python
print(f"Electricity anomaly : {len(an_idx_list)}")
```

```python
elec_df["label"]=0
for index in an_idx_list:
    elec_df.at[index,'label']=1
elec_df.to_csv('./Electricity_WHE_labeled.csv')
```
